```python
df1_3 = pd.read_csv('dataset_ex_3.csv')
df1_3 = df1_3.drop(columns=['DateTime'])

df1_3.head()
```

```python
# 1. Renomeie as três variáveis de consumo para Consumo_Zona_1, Consumo_Zona_2 e Consumo_Zona_3.
df1_3 = df1_3.rename(columns={
    "Temperature": "Temperatura",
    "Humidity": "Humidade",
    "Zone 1 Power Consumption": "Consumo_Zona_1",
    "Zone 2  Power Consumption": "Consumo_Zona_2",
    "Zone 3  Power Consumption": "Consumo_Zona_3",
    "Wind Speed": "Vel_Vento"})
df1_3.head()
```

```python
# 2. Determine o consumo máximo registrado em cada uma das três zonas
print(f'O maior consumo de eletrodomésticos na amostra Consumo_Zona_1 é de {df1_3["Consumo_Zona_1"].max():.2f} kWh')
print(f'O maior consumo de eletrodomésticos na amostra Consumo_Zona_2 é de {df1_3["Consumo_Zona_2"].max():.2f} kWh')
print(f'O maior consumo de eletrodomésticos na amostra Consumo_Zona_3 é de {df1_3["Consumo_Zona_3"].max():.2f} kWh')
```

```python
# 3. Identifique qual zona apresenta o maior pico de consumo na amostra.
# Seleciona apenas as colunas de consumo das zonas
colunas_de_consumo = ['Consumo_Zona_1', 'Consumo_Zona_2', 'Consumo_Zona_3']

# Encontra o valor máximo (pico) dentro de cada uma dessas colunas
picos_por_zona = df1_3[colunas_de_consumo].max()

# Descobre qual dessas colunas tem o maior valor geral usando idxmax()
zona_maior_pico = picos_por_zona.idxmax()
valor_do_pico = picos_por_zona.max()

print(f"A zona com o maior pico de consumo foi a {zona_maior_pico}, com o valor de {valor_do_pico:.2f} kWh")
```

```python
# 4. Para a zona identificada, calcule 70% do valor máximo e crie um DataFrame com os registros acima desse limiar.
limiar_70_df3 = df1_3[zona_maior_pico].max() * 0.7
df3_70 = df1_3[df1_3[zona_maior_pico] > limiar_70_df3]
df3_70
```

```python
# 5. Conte os registros e calcule o percentual que eles representam.
percentual_df3 = 100 * df3_70.shape[0] / df1_3.shape[0]
print(f"O número de consumidores com demanda acima de 70% é {df3_70.shape[0]} e o percentual da amostra que eles representam é de {percentual_df3:.2f}%")
```

```python
# 6. Calcule a temperatura média e crie um segundo DataFrame contendo consumo acima do limiar e temperatura acima da média.
media_temp_df3 = df1_3['Temperatura'].mean()
print(f"A temperatura média registrada na amostra é {media_temp_df3:.2f} °C")

df3_amb = df1_3[(df1_3[zona_maior_pico] > limiar_70_df3) & (df1_3['Temperatura'] > media_temp_df3)]
df3_amb
```

```python
# 7. Compare os dois conjuntos e explique o que ocorreu com a quantidade de registros após a inclusão da condição ambiental.
qtd_so_consumo = len(df3_70)
qtd_consumo_temp = len(df3_amb)
percentual_amb = 100 * qtd_consumo_temp / df1_3.shape[0]

print(f"Registros apenas com {zona_maior_pico} acima de 70% do máximo: {qtd_so_consumo}")
print(f"Registros com {zona_maior_pico} acima de 70% do máximo E temperatura acima da média: {qtd_consumo_temp}")
print(f"Esses {qtd_consumo_temp} registros representam {percentual_amb:.2f}% da amostra total")
```

Ao adicionar a temperatura acima da média como segundo critério, a quantidade de registros caiu de 2388 para 1597, ou seja, o conjunto ficou menor. Isso mostra que nem todos os momentos de consumo elevado na Consumo_Zona_1 acontecem quando a temperatura está acima da média: cerca de 33% dos registros de alto consumo ocorreram com temperatura abaixo da média. Ainda assim, a maior parte (aproximadamente 67%) dos picos de consumo coincide com temperaturas mais altas, o que é coerente com o contexto de Tétouan, já que temperaturas elevadas costumam aumentar o uso de sistemas de climatização e, consequentemente, o consumo elétrico. A condição ambiental funcionou como um filtro adicional, tornando o recorte mais restrito e mais específico do que o critério baseado apenas no consumo.

# **Dataset 5 — Wind & Solar Energy Production (Kaggle)**

# Etapa A

imagem_ex_5.svg

# Etapa B


```python
# 1. Renomeie as variáveis principais para Geracao_Solar e Geracao_Eolica.
df1_5 = pd.read_csv('dataset_ex_5.csv')
df_solar = df1_5[df1_5['Source'] == 'Solar']
df_eolica = df1_5[df1_5['Source'] == 'Wind']

df_solar = df_solar.rename(columns={'Production': 'Geracao_Solar'})
df_eolica = df_eolica.rename(columns={'Production': 'Geracao_Eolica'})

df1_5.head()
```

```python
# 2. Determine o valor máximo registrado para cada fonte
dfsolar_maior_valor = df_solar['Geracao_Solar'].max()
dfeolica_maior_valor = df_eolica['Geracao_Eolica'].max()

print(f"A maior potência solar registrada na amostra é de: {dfsolar_maior_valor} MWh")
print(f"A maior potência eólica registrada na amostra é de: {dfeolica_maior_valor} MWh")
```

```python
# 3. Calcule separadamente 70% do máximo solar e 70% do máximo eólico.
maximo70_solar = dfsolar_maior_valor * 0.7
maximo70_eolica = dfeolica_maior_valor * 0.7

print(f"O máximo solar é de: {maximo70_solar:.2f} MWh")
print(f"O máximo eólico é de: {maximo70_eolica:.2f} MWh")
```

```python
# 4. Crie um DataFrame para registros de alta geração solar e outro para registros de alta geração eólica.
df_alta_geracao_solar = df_solar[df_solar['Geracao_Solar'] > maximo70_solar]

df_alta_geracao_solar.head()
```

```python
df_alta_geracao_eolica = df_eolica[df_eolica['Geracao_Eolica'] > maximo70_eolica]

df_alta_geracao_eolica.head()
```

```python
# 5. Conte os registros de cada DataFrame e calcule seus respectivos percentuais.
qtd_alta_solar = len(df_alta_geracao_solar)
qtd_alta_eolica = len(df_alta_geracao_eolica)

perc_solar = (qtd_alta_solar / len(df_solar)) * 100
perc_eolica = (qtd_alta_eolica / len(df_eolica)) * 100

print(f"Registros de alta geração Solar: {qtd_alta_solar} ({perc_solar:.2f}%)")
print(f"Registros de alta geração Eólica: {qtd_alta_eolica} ({perc_eolica:.2f}%)")
```

```python
# 6. Compare qual fonte aparece com maior frequência acima de 70% do seu próprio máximo.

if perc_solar > perc_eolica:
    print(f"A fonte Solar aparece com maior frequência em alta geração ({perc_solar:.2f}%), superando a fonte Eólica ({perc_eolica:.2f}%).")
elif perc_eolica > perc_solar:
    print(f"A fonte Eólica aparece com maior frequência em alta geração ({perc_eolica:.2f}%), superando a fonte Solar ({perc_solar:.2f}%).")
else:
    print(f"As duas fontes aparecem com a mesma exata frequência de alta geração ({perc_solar:.2f}%).")
```

Ao analisar os percentuais, a fonte eólica aparece com maior frequência em períodos de alta geração. Ela registrou 2.87% dos seus registros acima da marca de 70% de sua capacidade máxima, enquanto a fonte solar registrou apenas 2.44%.

Não seria adequado utilizar um limite numérico único porque as duas fontes possuem capacidades de geração (escalas) completamente diferentes. Conforme calculado na Etapa 2, a usina eólica atinge picos muito maiores (22.929 MWh) do que a solar (16.316 MWh).  Se definíssemos um limite fixo de 15.000 MWh para as duas, por exemplo, estaríamos exigindo quase 92% da capacidade máxima da usina solar (um evento raro), mas apenas 65% da capacidade eólica (algo muito mais comum e fácil de atingir). Calcular o limite como 70% do próprio máximo normaliza a análise, garantindo que o conceito de 'alta geração' seja proporcional e justo para cada fonte.

Isso demonstra que absolutamente todos os episódios de demanda de potência extremamente alta geram, por consequência física, uma corrente elétrica muito superior à média geral, tornando essa segunda condição redundante para picos tão severos.
